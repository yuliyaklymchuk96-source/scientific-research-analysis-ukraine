```python

import pandas as pd
file_path = (
    "dataset_2026-08-22T18_36_56.791903119Z_"
    "DEFAULT_INTEGRATION_SSSU_DF_SCIENTIFIC_RESEARCH_"
    "DEVELOPMENT_LATEST.csv"
)

df = pd.read_csv(file_path)

print("Дані успішно завантажено!")
print("Кількість рядків:", df.shape[0])
print("Кількість колонок:", df.shape[1])

display(df.head())

```

```python
print("Назви колонок:\n")

for col in df.columns:
    print(col)

print("\nТипи даних:")
print(df.dtypes)
```

```python
missing_values = df.isna().sum()

print("Пропущені значення:")
print(missing_values)
```

```python
print(
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)
```

```python
duplicates = df.duplicated().sum()

print("Кількість повних дублікатів:", duplicates)
```

```python
df = df.drop_duplicates().copy()
```

```python
empty_columns = df.columns[
    df.isna().all()
].tolist()

print("Повністю порожні колонки:")
print(empty_columns)
```

```python
df = df.drop(columns=empty_columns)

print("Кількість колонок після видалення:", df.shape[1])
```

```python
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

print("Текстові колонки очищено.")
```

```python
df[text_columns] = df[text_columns].replace(
    {
        "": pd.NA,
        "NULL": pd.NA,
        "null": pd.NA,
        "N/A": pd.NA,
        "NA": pd.NA
    }
)
```

```python
year_columns = [
    str(year)
    for year in range(2016, 2026)
    if str(year) in df.columns
]

print("Роки:")
print(year_columns)
```

```python
for col in year_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print(df[year_columns].dtypes)
```

```python
print(
    df["Спеціальне значення даних"]
    .value_counts(dropna=False)
)
```

```python
print("========== РЕЗУЛЬТАТ ОЧИЩЕННЯ ==========")

print("Кількість рядків:", df.shape[0])
print("Кількість колонок:", df.shape[1])
print("Дублікатів:", df.duplicated().sum())

print("\nПропущені значення:")
print(
    df.isna()
    .sum()
    .sort_values(ascending=False)
)

display(df.head())
print(df.shape)
```

```python
output_file = "cleaned_research_expenses.csv"

df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("Очищення завершено!")
print("Файл збережено як:")
print(output_file)
```
